```python
##positive index
word= "youcando"
print(word[0])
print(word[1])
print(word[2])
print(word[3])
print(word[4])
```

```python
##negative index

word= "youcando"
print(word[-3])
print(word[-4])
```

```python
print("last character:",word[-1])
```

```python
#range slicing with step
#without step

print(word[0:6:1])
print(word[0:6])
print(word[0:7])
print(word[0:8])
```

```python
#SLICING with different steps

print(word[0:6:2])
print(word[0:8:2])
print(word[0:6:3])
print(word[0:8:3])
```
